```python
# Make sure the library is loaded
library(worldfootballR)

# Define the range of all seasons you want to get data for.
# According to the package data, seasons for Romania are available from 2006 to 2024.
all_seasons <- 2006:2024

# Loop through each season in the range
for(season_year in all_seasons) {

  # Use a tryCatch block to gracefully handle any potential errors for a specific year.
  # This ensures the loop won't stop if one year fails.
  tryCatch({

    # Print a message to show which season is being processed
    print(paste("--- Starting season:", season_year, "---"))

    # 1. Get the URLs for all teams for the current season in the loop
    team_urls <- tm_league_team_urls(country_name = "Romania", start_year = season_year)

    # 2. Get all transfers for that list of teams
    transfers <- tm_team_transfers(team_url = team_urls, transfer_window = "all")

    # 3. Create a unique file name for the current season
    file_name <- paste0("romania_transfers_", season_year, ".csv")

    # 4. Save the data for the current season to its CSV file
    write.csv(transfers, file_name, row.names = FALSE)

    # Print a success message for the current season
    print(paste("--- Successfully saved:", file_name, "---"))

  }, error = function(e) {
    # If an error occurs for a specific year, print the error message and continue to the next year
    print(paste("Could not fetch data for season", season_year, ". Error:", e$message))
  })
}

print("--- All seasons have been processed! ---")
```
